# Techno-Economic Periods: Timeline Visualizations

This notebook includes each period text in Markdown and one timeline graph per period.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

plt.rcParams['figure.figsize'] = (12, 2.8)
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
```

## 1. The Industrial Revolution (Age of Steam & Mechanics)
This first surge marked the transition from hand production methods to machines, new chemical manufacturing, and iron production processes.
Core Technologies: Mechanized cotton industry, wrought iron, water wheels.
Key Hub: Britain.

| Phase | Approx. Years | Description |
|---|---|---|
| Irruption | 1771-1780s | Begins with the opening of Arkwright's mill at Cromford (1771). The new technologies are introduced, offering huge cost reductions. |
| Frenzy | 1790s-1793 | Canal Mania. Intense investment in infrastructure (canals/turnpikes) to support the new paradigm. |
| The Turning Point | 1793-1797 | The Canal Panic. A period of recession and financial crash following the mania. |
| Synergy (Golden Age) | 1798-1810s | The full expansion of the factory system. The paradigm spreads across industries. |
| Maturity | 1810s-1829 | Market saturation. The technology reaches its limits; profitability declines, setting the stage for the next revolution. |

```python
phases = [
    ('Irruption', 1771, 1789, '#4E79A7'),
    ('Frenzy', 1790, 1793, '#F28E2B'),
    ('Turning Point', 1793, 1797, '#E15759'),
    ('Synergy', 1798, 1819, '#76B7B2'),
    ('Maturity', 1820, 1829, '#59A14F'),
]

fig, ax = plt.subplots()
for name, start, end, color in phases:
    ax.axvspan(start, end, color=color, alpha=0.45)
    ax.text((start + end) / 2, 0.5, name, ha='center', va='center', fontsize=9)

ax.set_title('1) Industrial Revolution (color bands)')
ax.set_xlim(1770, 1831)
ax.set_ylim(0, 1)
ax.set_yticks([])
ax.set_xlabel('Year')
plt.show()
```

## 2. The Age of Steam and Railways
The victory of the steam engine over water power and the massive build-out of rail networks created a true national market.
Core Technologies: Steam engines, railways, machine-made machines, postal service.
Key Hub: Britain (spreading to Europe and USA).
Shutterstock

| Phase | Approx. Years | Description |
|---|---|---|
| Irruption | 1829-1830s | Begins with the test of the "Rocket" steam engine for the Liverpool-Manchester line (1829). |
| Frenzy | 1840s-1847 | Railway Mania. Massive speculative investment in rail networks globally. |
| The Turning Point | 1848-1850 | The Railway Panic. Financial collapse followed by social unrest (Revolutions of 1848). |
| Synergy (Golden Age) | 1850-1860s | The Victorian Boom. The "Great British Leap." Railways lower the cost of transport, enabling mass markets. |
| Maturity | 1860s-1873 | Markets become saturated. Productivity gains slow down, leading to the "Great Depression" of 1873. |

```python
phases = [
    ('Irruption', 1829, 1839, '#1f77b4', '////'),
    ('Frenzy', 1840, 1847, '#ff7f0e', 'xxxx'),
    ('Turning Point', 1848, 1850, '#d62728', '....'),
    ('Synergy', 1851, 1869, '#2ca02c', 'oo'),
    ('Maturity', 1870, 1873, '#9467bd', '++'),
]

fig, ax = plt.subplots()
for name, start, end, color, hatch in phases:
    rect = patches.Rectangle((start, 0.15), end-start, 0.7, facecolor='white', edgecolor=color, hatch=hatch, linewidth=1.8)
    ax.add_patch(rect)
    ax.text((start + end) / 2, 0.5, name, ha='center', va='center', fontsize=9)

ax.set_title('2) Age of Steam & Railways (hatch patterns)')
ax.set_xlim(1828, 1874)
ax.set_ylim(0, 1)
ax.set_yticks([])
ax.set_xlabel('Year')
plt.show()
```

## 3. The Age of Steel, Electricity and Heavy Engineering
This era saw the rise of global markets, heavy chemistry, and the first major globalization of trade and finance.
Core Technologies: Bessemer steel, electricity, telegraph, transcontinental shipping, heavy chemicals.
Key Hubs: USA and Germany (overtaking Britain).
Getty Images

| Phase | Approx. Years | Description |
|---|---|---|
| Irruption | 1875-1880s | Begins with the opening of the Carnegie Bessemer steel plant in Pittsburgh (1875). |
| Frenzy | 1880s-1890s | Scramble for Africa, globalization of trade, and massive infrastructure projects (bridges, ships, tunnels). |
| The Turning Point | 1890-1895 | Baring Crisis (Argentina) and US Panics. A series of crashes shaking the global financial system. |
| Synergy (Golden Age) | 1895-1910s | The Belle Epoque (Europe) / Progressive Era (USA). Rapid economic growth and the rise of the great industrial powers. |
| Maturity | 1910s-1918 | The path to WWI. Market saturation and geopolitical tension lead to the end of the cycle. |

```python
phases = [
    ('Irruption', 1875, 1884, '#4E79A7'),
    ('Frenzy', 1885, 1890, '#F28E2B'),
    ('Turning Point', 1890, 1895, '#E15759'),
    ('Synergy', 1896, 1913, '#76B7B2'),
    ('Maturity', 1914, 1918, '#59A14F'),
]

fig, ax = plt.subplots()
for i, (name, start, end, color) in enumerate(phases):
    ax.axvspan(start, end, color=color, alpha=0.25 + i * 0.12)
    ax.text((start + end) / 2, 0.5, name, ha='center', va='center', fontsize=9)

ax.set_title('3) Steel, Electricity & Heavy Engineering (increasing intensity)')
ax.set_xlim(1874, 1919)
ax.set_ylim(0, 1)
ax.set_yticks([])
ax.set_xlabel('Year')
plt.show()
```

## 4. The Age of Oil, the Automobile and Mass Production
The era of standardization, mass consumption, and the rise of the middle class in the West.
Core Technologies: Internal combustion engine, oil, petrochemicals, assembly line, radio, plastics.
Key Hub: USA.

| Phase | Approx. Years | Description |
|---|---|---|
| Irruption | 1908-1920 | Begins with the launch of the Ford Model T (1908). Mass production methods challenge custom manufacturing. |
| Frenzy | 1920s | The Roaring Twenties. Speculation in stock markets, real estate, and radio utilities. |
| The Turning Point | 1929-1943 | The Great Crash of 1929 and the Great Depression. A long period of instability requiring massive institutional recomposition (New Deal, WWII). |
| Synergy (Golden Age) | 1944-1970s | The Post-War Golden Age. Suburbanization, mass consumption, and the welfare state. High growth and low inequality. |
| Maturity | 1970s | Stagflation. The Oil Crisis (1973) signals the limits of cheap energy and mass production efficiency. |

```python
phases = [
    ('Irruption', 1908, 1920, '#4E79A7'),
    ('Frenzy', 1920, 1929, '#F28E2B'),
    ('Turning Point', 1929, 1943, '#E15759'),
    ('Synergy', 1944, 1973, '#76B7B2'),
    ('Maturity', 1973, 1979, '#59A14F'),
]

fig, ax = plt.subplots()
y = 0.5
for name, start, end, color in phases:
    ax.plot([start, end], [y, y], color=color, linewidth=18, solid_capstyle='butt')
    ax.text((start + end) / 2, y + 0.2, name, ha='center', va='bottom', fontsize=9)

ax.scatter([1929, 1944, 1973], [y, y, y], color='black', s=18, zorder=3)
ax.set_title('4) Oil, Automobile & Mass Production (thick-segment line)')
ax.set_xlim(1907, 1980)
ax.set_ylim(0, 1)
ax.set_yticks([])
ax.set_xlabel('Year')
plt.show()
```

## 5. The Age of Information and Telecommunications
The current surge, defined by the microchip, digital networks, and the globalization of information.
Core Technologies: Microprocessors, internet, software, telecommunications, biotechnology.
Key Hub: USA (Silicon Valley).

| Phase | Approx. Years | Description |
|---|---|---|
| Irruption | 1971-1990s | Begins with the introduction of the Intel 4004 microprocessor (1971). The digital world begins to replace analog systems. |
| Frenzy | 1990s-2000 | The Dot-com Boom. Massive venture capital flows into internet startups and telecom infrastructure. |
| The Turning Point | 2000-? | The Dot-com Crash (2000) and the Financial Crisis (2008). According to Perez, we have been in a prolonged "Turning Point" phase, struggling to establish the institutions needed for the next Golden Age. |
| Synergy (Golden Age) | TBD | Potential Future Phase. If successful institutional adjustments occur (e.g., green growth, universal digital access), a sustainable global boom could follow. |
| Maturity | TBD | Future Phase. When the digital paradigm reaches its limits. |

```python
labels = ['Irruption\n1971-1990s', 'Frenzy\n1990s-2000', 'Turning Point\n2000-?', 'Synergy\nTBD', 'Maturity\nTBD']
colors = ['#4E79A7', '#F28E2B', '#E15759', '#76B7B2', '#59A14F']
hatches = ['', '//', '..', 'xx', 'oo']

fig, ax = plt.subplots(figsize=(12, 3.2))
for i, (label, color, hatch) in enumerate(zip(labels, colors, hatches)):
    rect = patches.Rectangle((i, 0.2), 1, 0.62, facecolor=color if i < 3 else 'white', edgecolor=color, hatch=hatch, linewidth=1.8, alpha=0.45 if i < 3 else 1)
    ax.add_patch(rect)
    ax.text(i + 0.5, 0.51, label, ha='center', va='center', fontsize=9)

ax.annotate('open-ended', xy=(2.98, 0.51), xytext=(3.25, 0.9), arrowprops={'arrowstyle': '->', 'lw': 1.1}, fontsize=9)
ax.set_xlim(0, 5)
ax.set_ylim(0, 1.05)
ax.set_xticks([])
ax.set_yticks([])
ax.set_title('5) Information & Telecommunications (conceptual timeline for open/TBD phases)')
plt.show()
```

```python
phase_colors = {
    'Irruption': '#4E79A7',
    'Frenzy': '#F28E2B',
    'Turning Point': '#E15759',
    'Synergy': '#76B7B2',
    'Maturity': '#59A14F',
}

period_hatches = {
    'Industrial Revolution': '////',
    'Steam & Railways': 'xxxx',
    'Steel/Electricity/Heavy Eng': '....',
    'Oil/Automobile/Mass Production': 'oo',
    'Information/Telecom': '++',
}

period_data = [
    (
        '1) Industrial Revolution',
        'Industrial Revolution',
        [
            ('Irruption', 1771, 1789, False),
            ('Frenzy', 1790, 1793, False),
            ('Turning Point', 1793, 1797, False),
            ('Synergy', 1798, 1819, False),
            ('Maturity', 1820, 1829, False),
        ],
    ),
    (
        '2) Steam & Railways',
        'Steam & Railways',
        [
            ('Irruption', 1829, 1839, False),
            ('Frenzy', 1840, 1847, False),
            ('Turning Point', 1848, 1850, False),
            ('Synergy', 1851, 1869, False),
            ('Maturity', 1870, 1873, False),
        ],
    ),
    (
        '3) Steel/Electricity/Heavy Eng',
        'Steel/Electricity/Heavy Eng',
        [
            ('Irruption', 1875, 1884, False),
            ('Frenzy', 1885, 1890, False),
            ('Turning Point', 1890, 1895, False),
            ('Synergy', 1896, 1913, False),
            ('Maturity', 1914, 1918, False),
        ],
    ),
    (
        '4) Oil/Automobile/Mass Production',
        'Oil/Automobile/Mass Production',
        [
            ('Irruption', 1908, 1920, False),
            ('Frenzy', 1920, 1929, False),
            ('Turning Point', 1929, 1943, False),
            ('Synergy', 1944, 1973, False),
            ('Maturity', 1973, 1979, False),
        ],
    ),
    (
        '5) Information/Telecom',
        'Information/Telecom',
        [
            ('Irruption', 1971, 1990, False),
            ('Frenzy', 1990, 2000, False),
            ('Turning Point', 2000, 2026, False),
            ('Synergy', 2027, 2038, True),
            ('Maturity', 2039, 2050, True),
        ],
    ),
]

fig, ax = plt.subplots(figsize=(15, 6.5))
bar_height = 0.62
y_positions = list(range(len(period_data) - 1, -1, -1))

for row_index, (row_label, hatch_key, phases) in enumerate(period_data):
    y = y_positions[row_index]
    hatch = period_hatches[hatch_key]

    for phase, start, end, is_projected in phases:
        rect = patches.Rectangle(
            (start, y - bar_height / 2),
            end - start,
            bar_height,
            facecolor=phase_colors[phase],
            alpha=0.45,
            edgecolor='black',
            linewidth=1.2,
            hatch=hatch,
        )
        ax.add_patch(rect)

        if is_projected:
            projection_outline = patches.Rectangle(
                (start, y - bar_height / 2),
                end - start,
                bar_height,
                fill=False,
                edgecolor='black',
                linewidth=1.8,
                linestyle='--',
            )
            ax.add_patch(projection_outline)

ax.set_title('Techno-Economic Revolutions: Combined Phase Timeline by Period')
ax.set_xlim(1770, 2050)
ax.set_ylim(-0.75, len(period_data) - 0.25)

xticks = sorted(set(list(range(1780, 2051, 20)) + [2026]))
ax.set_xticks(xticks)
ax.set_xlabel('Year')

ax.set_yticks(y_positions)
ax.set_yticklabels([row[0] for row in period_data])

ax.grid(axis='x', linestyle='--', linewidth=0.7, alpha=0.35)
ax.axvline(2026, color='black', linestyle=':', linewidth=1.2)
ax.text(2026 + 1, len(period_data) - 0.35, 'Present (2026)', fontsize=9, ha='left', va='bottom')

phase_handles = [
    patches.Patch(facecolor=color, edgecolor='black', alpha=0.45, label=phase)
    for phase, color in phase_colors.items()
]

period_handles = [
    patches.Patch(facecolor='white', edgecolor='black', hatch=hatch, label=period)
    for period, hatch in period_hatches.items()
]

phase_legend = ax.legend(
    handles=phase_handles,
    title='Phases',
    loc='upper left',
    bbox_to_anchor=(1.01, 1.0),
)
ax.add_artist(phase_legend)

ax.legend(
    handles=period_handles,
    title='Period Pattern',
    loc='upper left',
    bbox_to_anchor=(1.01, 0.45),
)

fig.tight_layout(rect=[0, 0, 0.8, 1])
plt.show()
```
