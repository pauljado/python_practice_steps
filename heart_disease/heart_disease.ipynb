```python
import glob
import pandas as pd
import os

#print(f"Current Directory: {os.getcwd()}")

file_list = glob.glob("*.csv")
print(f"Found {len(file_list)} files: {file_list}")

dfs = [pd.read_csv(f) for f in file_list]

df_heart = dfs[0]

print(df_heart.describe())


```

```python
import matplotlib.pyplot as plt
import seaborn as sns

ages = df_heart["Age"].nunique()
avg_age = df_heart["Age"].mean()

avg_cholesterol = df_heart["Cholesterol"].mean()

fig, ax = plt.subplots(2, 1, figsize=(7, 9))

sns.histplot(data=df_heart, x=df_heart["Age"], bins=ages, ax=ax[0])
ax[0].axvline(x=avg_age, color = "red", ls="--", lw=2, label="Average Age")
ax[0].set_title("Age distribution")
ax[0].legend()

sns.histplot(data=df_heart, x=df_heart["Cholesterol"], bins=50, ax=ax[1])
ax[1].axvline(x=avg_cholesterol, color = "red", ls="--", lw=2, label="Average Cholesterol")
ax[1].set_title("Cholesterol distribution")
ax[1].legend()


plt.tight_layout()
plt.show()
```
